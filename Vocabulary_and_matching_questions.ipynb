## Question 1
Use spaCy’s Matcher to find “cross-timber” written as crosstimber, cross timber or cross-timber. Name the rule CrossTimber.

```python
# build a matcher with three token patterns for cross-timber
import spacy
nlp=spacy.load('en_core_web_sm')
from spacy.matcher import Matcher
matcher=Matcher(nlp.vocab)
pattern1=[{'LOWER':'crosstimber'}]
pattern2=[{'LOWER':'cross'},{'LOWER':'timber'}]
pattern3=[{'LOWER':'cross'},{'IS_PUNCT':True},{'LOWER':'timber'}]
matcher.add('CrossTimber',[pattern1,pattern2,pattern3])
```

## Question 2          
Read the story from owlcreek.txt, apply the CrossTimber matcher, and print the matches.

```python
# read the story into a doc and print the matches found
with open('owlcreek.txt', 'r', encoding='cp1252') as f:
    doc3 = nlp(f.read())

found_matches=matcher(doc3)
print(found_matches)
```

## Question 3
Loop through the matches and print the match id, name, start, end and matched text.

```python
# loop through matches and print their details
for match_id,start,end in found_matches:
  string_id=nlp.vocab.strings[match_id]
  span=doc3[start:end]
  print(match_id,string_id,start,end,span.text)
```

## Question 4
Use spaCy’s PhraseMatcher to find these phrases: railroad bridge, cross-timber, Federal army, Owl Creek. Name it OwlCreek.

```python
# build a phrase matcher for the list of phrases
from spacy.matcher import PhraseMatcher
matcher=PhraseMatcher(nlp.vocab)
phrase_list=['railroad bridge','cross-timber','Federal army','Owl Creek']
phrase_patterns=[nlp(text) for text in phrase_list]
matcher.add('OwlCreek',None,*phrase_patterns)
```

## Question 5
Apply the phrase matcher to the story and print the matches.

```python
# run the phrase matcher on doc3 and print the matches
matches=matcher(doc3)
print(matches)
```
