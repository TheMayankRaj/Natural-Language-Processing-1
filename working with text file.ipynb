### 6. Word tokenization
Read `huggingface.txt` and tokenize the text.

```python
#6(d): removing punctuation after tokenization
doc=nlp(text)
for token in doc:
  if not token.is_punct:
    print(token.text)
```

### 8. Stop word removal using spacy
**(a)** Print every token of the text with `is_stop`.
**(b)** Print the stopwords present in the text.
**(c)** Remove the stopwords and punctuation and print the remaining words in a list.
**(d)** Remove the stopwords and punctuation, convert to lowercase and print the text.

```python
#8(b): stopwords present in the text
doc=nlp(text)
stopwords=[]


for token in doc:
  if token.is_stop:
    stopwords.append(token.text)

print(stopwords)
```

```python
#8(c): removing stopwords and punctuation
doc=nlp(text)
words=[]


for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.text)

print(words)
```

### 10. Stemming using PorterStemmer (with spacy)
**(a)** Print each alphabetic token of the text with its stem.
**(b)** Print each token in lowercase with its stem.
**(c)** Print the original text and the stemmed text.

```python
#10: spacy does not provide a built-in stemmer, so PorterStemmer of nltk is used
import spacy
from nltk.stem import PorterStemmer

nlp=spacy.load('en_core_web_sm')
stemmer=PorterStemmer()
```

```python
#10(a): stem of each alphabetic token
doc=nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
#10(b): stemming with lowercase
doc=nlp(text)
for token in doc:
  if token.is_alpha:
    word=token.text.lower()
    print(word,"->",stemmer.stem(word))
```

```python
#10(c): original text and stemmed text
doc=nlp(text)

stemmed_words=[]
for token in doc:
  if token.is_alpha:
    stemmed_words.append(stemmer.stem(token.text.lower()))

stemmed_text=" ".join(stemmed_words)
print('Original:')
print(text)

print("\nStemmed:")
print(stemmed_text)
```

```python
#11(f): lemmatization using nltk
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
words = [token.text.lower() for token in nlp(text) if token.is_alpha]
for word in words:
    print(word,"->",lemmatizer.lemmatize(word))
```

```python
#Q11(g): pos='v' for verbs
words = [token.text.lower() for token in nlp(text) if token.is_alpha]
for word in words:
    print(word,"->",lemmatizer.lemmatize(word,pos='v'))
```

```python
#Q11(h): pos='n' for nouns
words = [token.text.lower() for token in nlp(text) if token.is_alpha]
for word in words:
    print(word,"->",lemmatizer.lemmatize(word,pos='n'))
```

```python
#12(d): adjectives
doc=nlp(text)
for token in doc:
  if token.pos_ =='ADJ':
    print(token.text)
```

```python
#12(h): counting the POS tags
import spacy
from collections import Counter

doc=nlp(text)
pos_counts=Counter(token.pos_ for token in doc)
print(pos_counts)
```

```python
#12(i): bar graph of the POS tag counts
import spacy
from collections import Counter
import matplotlib.pyplot as plt

nlp=spacy.load('en_core_web_sm')

doc=nlp(text)
pos_counts=Counter(token.pos_ for token in doc)
plt.figure(figsize=(10,5))
plt.bar(pos_counts.keys(),
        pos_counts.values())

plt.xlabel('POS Tags')
plt.ylabel('frequency')
plt.title('POS Tag Frequency')

plt.show()
```
