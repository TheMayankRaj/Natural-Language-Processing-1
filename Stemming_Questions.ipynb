# Part A – NLTK Questions

## Question 1 (NLTK)
**Task:** Apply PorterStemmer on a list of verb forms and observe the stems.

```python
# Stem different forms of 'work'
words=['working','worked','works','worker','good']

for word in words:
  print(word, "->",stemmer.stem(word))
```

## Question 2 (NLTK)
**Task:** Stem every word of a simple sentence by first splitting the sentence.

```python
# Split a sentence and stem each word
text='The teachers are teaching different courses'

words=text.split()

for word in words:
  print(word, "->",stemmer.stem(word))
```

## Question 4 (NLTK)

## Question 5 (NLTK)
**Task:** Stem a list of plural nouns and observe how the stemmer handles them.

```python
# Stem plural nouns
words=['houses','cities','buses','boxes','teachers']

for word in words:
  print(word, "->",stemmer.stem(word))
```

## Question 6 (NLTK)
**Task:**  
1. Extract only alphabetic words from a sentence using regex  
2. Convert them to lowercase  
3. Apply stemming on each word  

```python
# Extract lowercase alphabetic words with regex and stem them
import re

text='Engineers are building robots and creating machine learning models'

words=re.findall(r'\b[a-zA-Z]+\b',text.lower())

for word in words:
  print(word, "->", stemmer.stem(word))
```

```python
# Import spaCy and PorterStemmer, load the model
import spacy
from nltk.stem import PorterStemmer

nlp=spacy.load('en_core_web_sm')
stemmer=PorterStemmer()
```

---
# Part B – spaCy + NLTK Questions

## Question 7 (spaCy + NLTK)
**Task:** Tokenize the sentence with spaCy and apply PorterStemmer only on alphabetic tokens.

```python
# Tokenize with spaCy and stem alphabetic tokens
text='the teachers are teaching'

doc=nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

## Question 8 (spaCy + NLTK)
**Task:** Stem different forms of the verb "play" that appear in multiple sentences.

```python
# Stem forms of 'play' across sentences
text='I play cricket.He plays cricket.They played yesterday.'

doc=nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

## Question 9 (spaCy + NLTK)
**Task:** Stem words related to "study" and print them in a neatly aligned format.

```python
# Stem tokens and print them in aligned columns

text = '''Students are studying different subjects.
They studied science and are studying mathematics.'''

doc = nlp(text)

for token in doc:
    if token.is_alpha:
        print(f"{token.text:15} -> {stemmer.stem(token.text)}")
```

## Question 10 (spaCy + NLTK)
**Task:**  
1. Tokenize the sentence with spaCy  
2. Keep only alphabetic tokens  
3. Convert them to lowercase and stem them  
4. Reconstruct the stemmed sentence  

```python
# Stem a sentence and rebuild it from the stemmed words
text='The children are playing games and enjoying themselves'

doc=nlp(text)

stemmed_words=[]
for token in doc:
  if token.is_alpha:
    stemmed_words.append(stemmer.stem(token.text.lower()))

stemmed_text=" ".join(stemmed_words)

print('Original:')
print(text)

print("\nStemmed:")
print(stemmed_text)
```
