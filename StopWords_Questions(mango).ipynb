# Part A – spaCy Questions

## Question 1 (spaCy)
**Task:** Check whether each word in the list is a stop word or not using `token.is_stop`.

```python
# Check whether each word is a stop word
words=['the','is','Java','and','Amazing']

for word in words:
  doc=nlp(word)

  for token in doc:
    print(token.text,"->",token.is_stop)
```

## Question 2 (spaCy)
**Task:** Print every token of a sentence along with its stop-word status.

```python
# Print each token with its stop-word status
text='Java is a very powerful programming language'

doc=nlp(text)

for token in doc:
    print(token.text,"->",token.is_stop)
```

## Question 3 (spaCy)
**Task:** Extract and print only the stop words present in the sentence.

```python
# Extract only the stop words from a sentence
text='This is a basic example of stopword filtering'

doc=nlp(text)
stopwords=[]

for token in doc:
  if token.is_stop:
    stopwords.append(token.text)

print(stopwords)
```

## Question 4 (spaCy)
**Task:** Remove stop words and punctuation from the sentence and keep only the meaningful words.

```python
# Remove stop words and punctuation
text='Java is an easy and powerful language'

doc=nlp(text)
words=[]

for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.text)

print(words)
```

## Question 5 (spaCy)
**Task:** Remove stop words and punctuation, convert remaining words to lowercase, and reconstruct the cleaned sentence.

```python
# Remove stop words and punctuation, lowercase, and rebuild the sentence
text='Java is a Powerful Programming Language'

doc=nlp(text)
result=[]

for token in doc:
  if not token.is_stop and not token.is_punct:
    result.append(token.text.lower())

print(" ".join(result))
```

# Part B – NLTK Questions

## Question 6 (NLTK)
**Task:** Check whether each word in the list is present in the NLTK English stop-word list.

```python
# Check whether each word is in the NLTK stop-word set
stop_words=set(stopwords.words('english'))

words=['the','is','Java','and','beautiful']

for word in words:
  print(word,"->",word.lower() in stop_words)
```

## Question 7 (NLTK)
**Task:** Extract and print only the stop words present in the sentence.

```python
# Extract only the stop words from a sentence
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop_words=set(stopwords.words('english'))

text='Java is a very useful programming language'

tokens=word_tokenize(text)
result=[]

for word in tokens:
  if word.lower() in stop_words:
    result.append(word)

print(result)
```

## Question 8 (NLTK)
**Task:** Remove stop words and punctuation from the sentence and keep only the meaningful words.

```python
# Remove stop words and punctuation from a sentence
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

stop_words=set(stopwords.words('english'))

text='The teachers are teaching data science'

tokens=word_tokenize(text)
result=[]

for word in tokens:
  if word.lower() not in stop_words and word not in string.punctuation:
    result.append(word)

print(result)
```

## Question 9 (NLTK)
**Task:** Remove stop words, convert remaining words to lowercase, and print the cleaned list.

```python
# Remove stop words and lowercase the remaining words
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop_words=set(stopwords.words('english'))

text='Java is a Powerful Programming Language'

tokens=word_tokenize(text)
result=[]

for word in tokens:
  if word.lower() not in stop_words:
    result.append(word.lower())

print(result)
```

## Question 10 (NLTK)
**Task:** Count how many stop words are present in the given sentence.

```python
# Count the stop words in a sentence
text='This is a simple example to demonstrate stopword removal'

tokens=word_tokenize(text)
count=0

for word in tokens:
  if word.lower() in stop_words:
    count+=1

print('Number of stopwords:',count)
```
