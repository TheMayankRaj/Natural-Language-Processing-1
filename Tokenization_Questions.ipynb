# Part A – NLTK Questions

## Question 1 (NLTK)
**Task:** Tokenize the given sentence into individual words using `word_tokenize`.

```python
# Tokenize a sentence into words
import nltk
from nltk.tokenize import word_tokenize

text='Machine learning is transforming the technology industry'

tokens=word_tokenize(text)

print(tokens)
```

## Question 2 (NLTK)
**Task:** Split the given paragraph into separate sentences using `sent_tokenize`.

```python
# Split a paragraph into sentences
from nltk.tokenize import sent_tokenize

text='Deep learning models require large datasets. Transformers have revolutionized NLP. Attention mechanisms improve model performance.'

sentences=sent_tokenize(text)

print(sentences)
```

## Question 3 (NLTK)
**Task:** Tokenize a multi-line text into words using `word_tokenize`.

```python
# Tokenize multi-line text into words
from nltk.tokenize import word_tokenize

text=""" Deep learning is a subset of machine learning.
Neural networks power modern AI systems.
Tokenization is the first step in text processing."""

tokens=word_tokenize(text)

print(tokens)
```

## Question 4 (NLTK)
**Task:** Tokenize a sentence that contains punctuation marks (comma, question mark, exclamation mark) and observe how NLTK treats them.

```python
# Tokenize a sentence with punctuation
from nltk.tokenize import word_tokenize

text='Hello, what is your name? I am learning NLP!'

tokens=word_tokenize(text)

print(tokens)
```

## Question 5 (NLTK)
**Task:**  
1. Tokenize the text  
2. Remove punctuation marks  
3. Convert all tokens to lowercase  
4. Count the number of remaining words  

```python
# Tokenize, remove punctuation, lowercase, and count words
import string
from nltk.tokenize import word_tokenize

text="Artificial intelligence is powerful and artificial intelligence is everywhere."

tokens=word_tokenize(text)

words=[token for token in tokens if token not in string.punctuation]

lower_words=[word.lower() for word in words]

print('All tokens:',tokens)
print('Words without punctuation:',words)
print('Lowercase words:',lower_words)
print('Number of words: ',len(words))
```

# Part B – spaCy Questions

## Question 6 (spaCy)
**Task:** Load the spaCy model and tokenize the given sentence. Print each token on a new line.

```python
# Print each spaCy token on a new line
text='Natural language processing is fascinating'

doc=nlp(text)

for token in doc:
  print(token.text)
```

## Question 7 (spaCy)
**Task:** Tokenize the sentence and print the **index** of each token along with the token text.

```python
# Print each token with its index
text='Natural language processing is fascinating.'

doc=nlp(text)

for token in doc:
  print(token.i,token.text)
```

## Question 8 (spaCy)
**Task:** Examine important token attributes (`is_alpha`, `is_punct`, `is_space`) for the given sentence.

```python
# Print token attributes (alpha, punct, space)
text='Natural language processing is fascinating.'

doc=nlp(text)

for token in doc:
  print(
      token.text,
      token.is_alpha,
      token.is_punct,
      token.is_space )
```

## Question 9 (spaCy)
**Task:** Tokenize a sentence that contains an **email address** and a **URL**. Observe how spaCy treats them as single tokens.

```python
# Tokenize text containing an email and a URL
text='Please contact support@company.org or visit https://www.company.org for details.'

doc=nlp(text)

for token in doc:
  print(token.text)
```

## Question 10 (spaCy)
**Task:**  
1. Process the text sentence by sentence  
2. Print each sentence and its tokens  
3. Display a table of token attributes (`is_alpha`, `like_num`, `is_punct`)  

```python
# Print sentences, their tokens, and an attribute table
text='Natural language processing is easy to learn. It is widely used in Data Science and AI.'

doc=nlp(text)

for sentence in doc.sents:
  print('Sentence:',sentence.text)
  for token in sentence:
   print(token.text)

print()

print('Token\tAlpha\tNumber\tPunctuation')
for token in doc:
  print(token.text,"\t",token.is_alpha,"\t",token.like_num,"\t",token.is_punct)
```
