## Question 1 (spaCy)
**Task:** Lemmatize every token of the given sentence and print the original word with its lemma.

```python
# Print each token with its spaCy lemma
text='the girls are dancing in the park.'

doc=nlp(text)

for token in doc:
  print(token.text, "->", token.lemma_)
```

## Question 5 (spaCy)
**Task:**  
1. Lemmatize the sentence  
2. Remove stop words and punctuation  
3. Keep only the useful lemmas  

```python
# Lemmatize and keep only non-stop, non-punctuation words
text='The teachers are teaching computer science'

doc=nlp(text)

words=[]
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.lemma_)

print(words)
```

## Question 10 (NLTK)
**Task:**  
1. Tokenize a sentence  
2. Automatically detect POS using `pos_tag`  
3. Map the POS tag to WordNet format  
4. Lemmatize each alphabetic word with the correct POS  

```python
# Lemmatize a sentence using automatically detected POS tags
import nltk
from nltk import word_tokenize, pos_tag
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer

# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

# Map POS tags to WordNet format
def get_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

text = "The children are playing with their toys."

# Tokenize the sentence
tokens = word_tokenize(text)

# Detect POS tags
tagged = pos_tag(tokens)

# Lemmatize each word
for word, tag in tagged:
    if word.isalpha():
        lemma = lemmatizer.lemmatize(word, pos=get_pos(tag))
        print(word, "-", lemma)
```
